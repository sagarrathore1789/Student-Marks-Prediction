```python
import pandas as pd

df = pd.read_csv(r"D:\AI_ML\Student-Marks-Prediction\dataset\student_data.csv")
print("Shape:", df.shape)
print("\nDescribe:\n", df.describe().round(2))
print("\nMissing:\n", df.isnull().sum())
print("\nFirst 5 rows:\n", df.head())
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 4, figsize=(18, 4))
features = ["study_hours", "attendance", "previous_marks", "assignment_scores"]

for ax, col in zip(axes, features):
    ax.scatter(df[col], df["final_marks"], alpha=0.6)
    ax.set_xlabel(col)
    ax.set_ylabel("final_marks")
    ax.set_title(f"{col} vs final_marks")

plt.tight_layout()
plt.show()
```
